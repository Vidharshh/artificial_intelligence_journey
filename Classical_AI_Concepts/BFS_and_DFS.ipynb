# Search Techniques

Purpose:  Find a path from an initial state to a goal state.

Imagine a maze. You know where you are starting and where you need to go, but you don't know which route will get you there.

Search algorithms systematically explore possible paths.

## Breadth First Search Algorithm


          A
        /   \
       B     C
      / \   / \
     D   E F   G


Let us say tha above is the graph

BFS explores:

Step 1: Visit A.

Step 2: Visit A's immediate neighbors: B, C.

Step 3: Visit the next level: D, E, F, G.

Traversal order is : A → B → C → D → E → F → G

Geeksforgeeks link : https://www.geeksforgeeks.org/dsa/breadth-first-search-or-bfs-for-a-graph/$0

The algorithm explores one complete level before moving to the next.


Where is BFS used?
- Finding shortest paths in unweighted graphs.
- Social networks: finding connections by degree of separation.
- Web crawling and graph traversal.
- Exploring neighboring states in search problems.
- Finding the minimum number of moves in puzzles.



```python
from collections import deque

# traversal function

def bfs(graph, start):
    queue = deque([start])
    visited = {start}

    while queue:
        node = queue.popleft()
        print(node, end=" ")

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

# Finding shortest path through BFS
from collections import deque

def bfs_shortest_path(graph, start, target):
    queue = deque([[start]])
    visited = {start}

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node == target:
            return path

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(path + [neighbor])

    return None

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}
```

```python
bfs(graph, 'A')
```

```python
bfs(graph,'B')
```

```python
print(bfs_shortest_path(graph, 'A', 'E'))
```

# Depth-First Search Algorithm

          A
        /   \
       B     C
      / \   / \
     D   E F   G

Assume we explore the left neighbor first.

Step 1: Start at A.

Step 2: Go to B.

Step 3: Go deeper to D.

Step 4: D has no unvisited neighbors, so backtrack to B.

Step 5: Visit E.

Step 6: Backtrack to A and explore C, followed by F and G.

Traversal order: A → B → D → E → C → F → G

DFS explores one branch as deeply as possible before backtracking to explore another branch.

Imagine exploring a maze. You keep walking down one corridor until you reach a dead end. Then you return to the last junction and try another corridor.

That's DFS.

Geeksforgeeks link : https://www.geeksforgeeks.org/dsa/depth-first-search-or-dfs-for-a-graph/$0

Where is DFS used?
- Maze solving.
- Detecting cycles in graphs.
- Topological sorting.
- Finding connected components.
- Dependency analysis.
- Exploring state spaces.
- Backtracking problems.


```python
# traversal function

def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)
    print(start, end=" ")

    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)

# finding a path through dfs

def dfs_path(graph, start, target, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)

    if start == target:
        return [start]

    for neighbor in graph[start]:
        if neighbor not in visited:
            path = dfs_path(graph, neighbor, target, visited)

            if path is not None:
                return [start] + path

    return None
```

```python
dfs(graph, 'A')
```

```python
dfs(graph, 'B')
```

```python

print(dfs_path(graph, 'A', 'F'))
```
